# **Transferable Data to Work in Another Notebook**
- Group data.
- Conduct a spatial analysis.
- Prepare a GEOJSON for transfer.to my working notebook.

```python
#Now the analysis begins
#This is using the groupby function to create a unique station count for each individual station across each year
#Call the new column Trip Count
station_year_counts = (
    lyft_gdf_copy.groupby(["Station ID", "Station Name","City","Year"])
    .size()
    .reset_index(name="Trip Count")
)
```

```python
#Lets inspect.
#Looks pretty good to me
#I actually had to fix this but deleted the old code. Basically, I was grouping by geography but the slight changes in coordinates was creating a bunch of unique values so I just used station name instead
station_year_counts
```

```python
#Need to add coordinates back into this data
#Using median gives me the average of all the unique station values so I have one, consistent reference point
station_coords = (
    lyft_gdf.groupby(["Station ID", "Station Name"])
    .agg({
        "Latitude": "median",
        "Longitude": "median"
    })
    .reset_index()
)
```

```python
#merge them back together so I can have all the data under one variable
station_year_counts = station_year_counts.merge(
    station_coords,
    on=["Station ID", "Station Name"],
    how="left"
)
```

```python
#Everything looks really good!
station_year_counts
```

```python
#Now I am turning into a gdf for analysis
station_year_gdf = gpd.GeoDataFrame(
    station_year_counts,
    geometry=gpd.points_from_xy(
        station_year_counts["Longitude"],
        station_year_counts["Latitude"]
    ),
    crs="EPSG:4326"
)
```
